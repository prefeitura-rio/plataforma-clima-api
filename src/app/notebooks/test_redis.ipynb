```python
import json
import numpy as np
from datetime import datetime


log = print


def save_on_redis(
    redis_client,
    values,
    redis_hash: str = None,
    redis_key: str = None,
    keep_last: int = 50,
    sort_key: str = None,
) -> None:
    """
    Save values on redis. If values are a list, order ir and keep only last names.
    """
    print(values)
    if isinstance(values, list):
        # try:
        #     values = list(set(values))
        # except TypeError:
        #     log("I was not possible to use set() to filter duplicates on this list.")
        # values = sorted(values, key=sort_key)
        log(f"\n\nBefore keep_last {values}\n\n")
        values = values[-keep_last:]

    if isinstance(values, dict):
        values = json.dumps(values)
    values = json.dumps(values)

    log(f"Saving files {values} on redis {redis_hash} {redis_key}")

    if redis_hash and redis_key:
        redis_client.hset(redis_hash, redis_key, values)
    elif redis_key:
        redis_client.set(redis_key, values)
    log(f"Saved to Redis hash: {redis_hash}, key: {redis_key}, value: {values}")


def sort_list_by_dict_key(item, key: str = "timestamp") -> datetime:
    """
    Sorts a list of dictionaries based on a specified key.

    Parameters:
    item (dict): A dictionary inside a list to be sorted.
    key (str, optional): The key to sort the list by. Defaults to "timestamp".

    Returns:
    datetime.datetime: The sorted datetime object based on the specified key.
    """
    return datetime.fromisoformat(item[key])


point_values_updated = [
    [
        {"timestamp": "2025-01-09 09:00:20", "valor": np.nan},
        {"timestamp": "2025-01-09 09:10:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:20:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:30:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:40:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:50:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:00:20", "valor": 0.19902298821428568},
        {"timestamp": "2025-01-09 10:10:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:20:20", "valor": 0.05370461586734693},
        {"timestamp": "2025-01-09 10:30:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:40:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:50:20", "valor": 0.0},
    ]
]
save_on_redis(
    redis_client,
    values=point_values_updated,
    redis_key="cp",
    keep_last=12,
    sort_key=sort_list_by_dict_key,
)
```

```python
import numpy as np
import pickle

print("output before", redis_client.get("cape"))

point_values_updated = [
    [
        {"timestamp": "2025-01-09 09:00:20", "valor": np.nan},
        {"timestamp": "2025-01-09 09:10:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:20:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:30:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:40:20", "valor": 0.0},
        {"timestamp": "2025-01-09 09:50:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:00:20", "valor": 0.19902298821428568},
        {"timestamp": "2025-01-09 10:10:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:20:20", "valor": 0.05370461586734693},
        {"timestamp": "2025-01-09 10:30:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:40:20", "valor": 0.0},
        {"timestamp": "2025-01-09 10:50:20", "valor": 0.0},
    ]
]

# Serialize the data
serialized_data = pickle.dumps(point_values_updated)

# Save the serialized data to Redis
redis_client.set("cape", serialized_data)

# Retrieve the serialized data from Redis
retrieved_data = redis_client.get("cape")

# Deserialize the data
deserialized_data = pickle.loads(retrieved_data)

print("after", deserialized_data)
```
